```python
# ============================================================
# HELP DESK TICKET SYSTEM - SQL PORTFOLIO PROJECT
# Google Colab + SQLite + Pandas + Matplotlib
# ============================================================

import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

# ------------------------------------------------------------
# 1. CONNECT TO SQLITE DATABASE
# ------------------------------------------------------------

conn = sqlite3.connect("help_desk_ticket_system.db")
cursor = conn.cursor()

print("Database connected successfully.\n")


# ------------------------------------------------------------
# 2. DROP OLD TABLES IF THEY EXIST
# This lets you rerun the notebook without errors.
# ------------------------------------------------------------

cursor.executescript("""
DROP TABLE IF EXISTS ticket_updates;
DROP TABLE IF EXISTS tickets;
DROP TABLE IF EXISTS categories;
DROP TABLE IF EXISTS technicians;
DROP TABLE IF EXISTS users;
DROP TABLE IF EXISTS departments;
""")


# ------------------------------------------------------------
# 3. CREATE TABLES
# ------------------------------------------------------------

cursor.executescript("""
CREATE TABLE departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
);

CREATE TABLE users (
    user_id INTEGER PRIMARY KEY,
    first_name TEXT NOT NULL,
    last_name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    department_id INTEGER,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
);

CREATE TABLE technicians (
    technician_id INTEGER PRIMARY KEY,
    technician_name TEXT NOT NULL,
    skill_level TEXT NOT NULL
);

CREATE TABLE categories (
    category_id INTEGER PRIMARY KEY,
    category_name TEXT NOT NULL
);

CREATE TABLE tickets (
    ticket_id INTEGER PRIMARY KEY,
    user_id INTEGER NOT NULL,
    technician_id INTEGER,
    category_id INTEGER NOT NULL,
    issue_description TEXT NOT NULL,
    priority TEXT NOT NULL,
    status TEXT NOT NULL,
    created_date TEXT NOT NULL,
    resolved_date TEXT,
    FOREIGN KEY (user_id)
        REFERENCES users(user_id),
    FOREIGN KEY (technician_id)
        REFERENCES technicians(technician_id),
    FOREIGN KEY (category_id)
        REFERENCES categories(category_id)
);

CREATE TABLE ticket_updates (
    update_id INTEGER PRIMARY KEY,
    ticket_id INTEGER NOT NULL,
    update_note TEXT NOT NULL,
    update_date TEXT NOT NULL,
    FOREIGN KEY (ticket_id)
        REFERENCES tickets(ticket_id)
);
""")

conn.commit()

print("Tables created successfully.\n")


# ------------------------------------------------------------
# 4. INSERT DEPARTMENTS
# ------------------------------------------------------------

departments = [
    (1, "Finance"),
    (2, "Human Resources"),
    (3, "Marketing"),
    (4, "Operations"),
    (5, "Sales"),
    (6, "Executive")
]

cursor.executemany("""
INSERT INTO departments
(department_id, department_name)
VALUES (?, ?)
""", departments)


# ------------------------------------------------------------
# 5. INSERT USERS
# ------------------------------------------------------------

users = [
    (1, "Ashley", "Brown", "ashley.brown@company.com", 1),
    (2, "Michael", "Johnson", "michael.johnson@company.com", 2),
    (3, "Sarah", "Williams", "sarah.williams@company.com", 3),
    (4, "Daniel", "Smith", "daniel.smith@company.com", 4),
    (5, "Emily", "Davis", "emily.davis@company.com", 5),
    (6, "James", "Wilson", "james.wilson@company.com", 1),
    (7, "Olivia", "Taylor", "olivia.taylor@company.com", 4),
    (8, "David", "Anderson", "david.anderson@company.com", 5),
    (9, "Sophia", "Thomas", "sophia.thomas@company.com", 3),
    (10, "Robert", "Jackson", "robert.jackson@company.com", 6)
]

cursor.executemany("""
INSERT INTO users
(user_id, first_name, last_name, email, department_id)
VALUES (?, ?, ?, ?, ?)
""", users)


# ------------------------------------------------------------
# 6. INSERT TECHNICIANS
# ------------------------------------------------------------

technicians = [
    (1, "Kevin Martinez", "Level 1"),
    (2, "Rachel Green", "Level 1"),
    (3, "Marcus Lee", "Level 2"),
    (4, "Jessica Carter", "Senior"),
    (5, "Brian Adams", "Level 2")
]

cursor.executemany("""
INSERT INTO technicians
(technician_id, technician_name, skill_level)
VALUES (?, ?, ?)
""", technicians)


# ------------------------------------------------------------
# 7. INSERT CATEGORIES
# ------------------------------------------------------------

categories = [
    (1, "Password Reset"),
    (2, "Network / Wi-Fi"),
    (3, "Software Issue"),
    (4, "Hardware Issue"),
    (5, "Email Issue"),
    (6, "Account Access"),
    (7, "Printer Issue"),
    (8, "Security Incident")
]

cursor.executemany("""
INSERT INTO categories
(category_id, category_name)
VALUES (?, ?)
""", categories)


# ------------------------------------------------------------
# 8. INSERT TICKETS
# ------------------------------------------------------------

tickets = [
    (1, 1, 1, 1, "User forgot Windows password.", "Low", "Resolved", "2026-09-01 08:10:00", "2026-09-01 08:35:00"),
    (2, 2, 2, 2, "Laptop will not connect to office Wi-Fi.", "High", "Resolved", "2026-09-01 09:20:00", "2026-09-01 11:00:00"),
    (3, 3, 3, 3, "Microsoft Excel keeps crashing.", "Medium", "Resolved", "2026-09-02 10:00:00", "2026-09-02 14:20:00"),
    (4, 4, 4, 4, "Laptop screen is flickering.", "High", "Resolved", "2026-09-02 11:15:00", "2026-09-03 10:00:00"),
    (5, 5, 1, 5, "Unable to receive company emails.", "High", "Resolved", "2026-09-03 08:45:00", "2026-09-03 10:30:00"),
    (6, 6, 2, 6, "Employee account locked after failed login attempts.", "Medium", "Resolved", "2026-09-03 13:00:00", "2026-09-03 13:40:00"),
    (7, 7, 3, 7, "Department printer is offline.", "Low", "Resolved", "2026-09-04 09:30:00", "2026-09-04 11:10:00"),
    (8, 8, 4, 8, "Suspicious phishing email received.", "Critical", "Resolved", "2026-09-04 10:20:00", "2026-09-04 11:00:00"),
    (9, 9, 5, 2, "VPN connection disconnects repeatedly.", "High", "Resolved", "2026-09-05 14:00:00", "2026-09-05 17:10:00"),
    (10, 10, 4, 3, "PowerPoint will not open.", "Medium", "Resolved", "2026-09-06 08:30:00", "2026-09-06 10:00:00"),

    (11, 1, 1, 2, "Internet connection is extremely slow.", "Medium", "Resolved", "2026-09-07 08:15:00", "2026-09-07 12:00:00"),
    (12, 2, 2, 1, "Password reset required.", "Low", "Resolved", "2026-09-07 09:40:00", "2026-09-07 10:00:00"),
    (13, 3, 3, 5, "Outlook search function is not working.", "Medium", "Resolved", "2026-09-08 11:00:00", "2026-09-08 15:20:00"),
    (14, 4, 5, 4, "Desktop computer does not power on.", "Critical", "Resolved", "2026-09-08 12:15:00", "2026-09-08 14:30:00"),
    (15, 5, 1, 6, "User cannot access shared drive.", "High", "Resolved", "2026-09-09 09:00:00", "2026-09-09 11:45:00"),
    (16, 6, 2, 3, "Adobe Acrobat freezes when opening PDFs.", "Low", "Resolved", "2026-09-09 13:10:00", "2026-09-09 16:00:00"),
    (17, 7, 3, 2, "No network connection in conference room.", "High", "Resolved", "2026-09-10 08:05:00", "2026-09-10 09:50:00"),
    (18, 8, 4, 8, "Malware alert detected on workstation.", "Critical", "Resolved", "2026-09-10 10:15:00", "2026-09-10 12:00:00"),
    (19, 9, 5, 7, "Printer is printing blank pages.", "Medium", "Resolved", "2026-09-11 09:25:00", "2026-09-11 11:00:00"),
    (20, 10, 4, 5, "Unable to send email attachments.", "Medium", "Resolved", "2026-09-11 13:00:00", "2026-09-11 15:30:00"),

    (21, 1, 1, 1, "Password expired and user cannot log in.", "Low", "Resolved", "2026-09-12 08:00:00", "2026-09-12 08:20:00"),
    (22, 2, 2, 4, "Keyboard keys are not responding.", "Medium", "Resolved", "2026-09-12 10:00:00", "2026-09-12 12:30:00"),
    (23, 3, 3, 3, "Teams application crashes during meetings.", "High", "Resolved", "2026-09-13 09:15:00", "2026-09-13 11:45:00"),
    (24, 4, 5, 2, "Ethernet connection not working.", "High", "Resolved", "2026-09-13 10:30:00", "2026-09-13 13:00:00"),
    (25, 5, 1, 6, "Permission denied when accessing CRM system.", "High", "Resolved", "2026-09-14 08:45:00", "2026-09-14 10:30:00"),
    (26, 6, 2, 7, "Printer queue is stuck.", "Low", "Resolved", "2026-09-14 12:00:00", "2026-09-14 13:20:00"),
    (27, 7, 3, 5, "Mailbox is full.", "Medium", "Resolved", "2026-09-15 09:30:00", "2026-09-15 10:45:00"),
    (28, 8, 4, 8, "Multiple failed login attempts detected.", "Critical", "Resolved", "2026-09-15 11:00:00", "2026-09-15 12:10:00"),
    (29, 9, 5, 3, "CRM application running very slowly.", "Medium", "In Progress", "2026-09-16 08:30:00", None),
    (30, 10, 4, 2, "Executive laptop cannot connect to VPN.", "Critical", "In Progress", "2026-09-16 09:00:00", None),

    (31, 1, 1, 4, "Laptop battery drains quickly.", "Low", "Open", "2026-09-17 10:00:00", None),
    (32, 2, 2, 5, "Outlook calendar not syncing.", "Medium", "Open", "2026-09-18 08:30:00", None),
    (33, 3, 3, 6, "Cannot access marketing shared folder.", "High", "In Progress", "2026-09-18 09:45:00", None),
    (34, 4, 5, 7, "Warehouse printer disconnected.", "Medium", "Open", "2026-09-19 07:30:00", None),
    (35, 5, 1, 3, "Sales software fails to launch.", "High", "In Progress", "2026-09-19 10:15:00", None),
    (36, 6, 2, 1, "Forgot network password.", "Low", "Resolved", "2026-09-20 08:00:00", "2026-09-20 08:15:00"),
    (37, 7, 3, 2, "Wi-Fi unavailable on second floor.", "High", "Open", "2026-09-20 11:00:00", None),
    (38, 8, 4, 8, "Employee clicked suspicious link.", "Critical", "In Progress", "2026-09-21 09:20:00", None),
    (39, 9, 5, 4, "Monitor will not turn on.", "Medium", "Open", "2026-09-21 12:10:00", None),
    (40, 10, 4, 6, "Executive cannot access finance dashboard.", "Critical", "Open", "2026-09-22 07:50:00", None)
]

cursor.executemany("""
INSERT INTO tickets
(
    ticket_id,
    user_id,
    technician_id,
    category_id,
    issue_description,
    priority,
    status,
    created_date,
    resolved_date
)
VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
""", tickets)


# ------------------------------------------------------------
# 9. INSERT TICKET UPDATES
# ------------------------------------------------------------

updates = [
    (1, 2, "Technician restarted network adapter and reconnected Wi-Fi.", "2026-09-01 10:15:00"),
    (2, 4, "Hardware diagnostics completed. Display cable replaced.", "2026-09-03 09:20:00"),
    (3, 8, "Phishing email quarantined and sender blocked.", "2026-09-04 10:45:00"),
    (4, 14, "Faulty power supply identified and replaced.", "2026-09-08 13:40:00"),
    (5, 18, "Workstation isolated and malware removed.", "2026-09-10 11:00:00"),
    (6, 28, "Account temporarily locked and password reset.", "2026-09-15 11:40:00"),
    (7, 29, "Application logs collected for further analysis.", "2026-09-16 10:00:00"),
    (8, 30, "VPN configuration under review.", "2026-09-16 10:30:00"),
    (9, 33, "Folder permissions escalated to Level 2 support.", "2026-09-18 11:00:00"),
    (10, 35, "Software reinstall scheduled.", "2026-09-19 13:00:00"),
    (11, 38, "Endpoint scan started and user password reset.", "2026-09-21 10:00:00")
]

cursor.executemany("""
INSERT INTO ticket_updates
(update_id, ticket_id, update_note, update_date)
VALUES (?, ?, ?, ?)
""", updates)

conn.commit()

print("Sample data inserted successfully.\n")


# ------------------------------------------------------------
# 10. HELPER FUNCTION FOR SQL QUERIES
# ------------------------------------------------------------

def run_query(title, query):
    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)

    df = pd.read_sql_query(query, conn)

    display(df)

    return df


# ------------------------------------------------------------
# 11. VIEW ALL TICKETS
# ------------------------------------------------------------

all_tickets = run_query(
    "1. ALL HELP DESK TICKETS",
    """
    SELECT *
    FROM tickets;
    """
)


# ------------------------------------------------------------
# 12. TICKET COUNT BY STATUS
# ------------------------------------------------------------

status_summary = run_query(
    "2. TICKET COUNT BY STATUS",
    """
    SELECT
        status,
        COUNT(*) AS ticket_count
    FROM tickets
    GROUP BY status
    ORDER BY ticket_count DESC;
    """
)


# ------------------------------------------------------------
# 13. TICKET COUNT BY PRIORITY
# ------------------------------------------------------------

priority_summary = run_query(
    "3. TICKET COUNT BY PRIORITY",
    """
    SELECT
        priority,
        COUNT(*) AS ticket_count
    FROM tickets
    GROUP BY priority
    ORDER BY ticket_count DESC;
    """
)


# ------------------------------------------------------------
# 14. MOST COMMON IT ISSUES
# ------------------------------------------------------------

category_summary = run_query(
    "4. MOST COMMON IT ISSUE CATEGORIES",
    """
    SELECT
        c.category_name,
        COUNT(t.ticket_id) AS ticket_count
    FROM tickets t
    JOIN categories c
        ON t.category_id = c.category_id
    GROUP BY c.category_name
    ORDER BY ticket_count DESC;
    """
)


# ------------------------------------------------------------
# 15. DEPARTMENTS SUBMITTING THE MOST TICKETS
# ------------------------------------------------------------

department_summary = run_query(
    "5. DEPARTMENTS WITH THE MOST SUPPORT TICKETS",
    """
    SELECT
        d.department_name,
        COUNT(t.ticket_id) AS ticket_count
    FROM tickets t
    JOIN users u
        ON t.user_id = u.user_id
    JOIN departments d
        ON u.department_id = d.department_id
    GROUP BY d.department_name
    ORDER BY ticket_count DESC;
    """
)


# ------------------------------------------------------------
# 16. TECHNICIAN WORKLOAD
# ------------------------------------------------------------

technician_workload = run_query(
    "6. TECHNICIAN WORKLOAD",
    """
    SELECT
        tech.technician_name,
        tech.skill_level,
        COUNT(t.ticket_id) AS assigned_tickets
    FROM technicians tech
    LEFT JOIN tickets t
        ON tech.technician_id = t.technician_id
    GROUP BY
        tech.technician_name,
        tech.skill_level
    ORDER BY assigned_tickets DESC;
    """
)


# ------------------------------------------------------------
# 17. TECHNICIANS WITH MOST RESOLVED TICKETS
# ------------------------------------------------------------

resolved_by_technician = run_query(
    "7. TECHNICIANS WITH THE MOST RESOLVED TICKETS",
    """
    SELECT
        tech.technician_name,
        COUNT(t.ticket_id) AS resolved_tickets
    FROM tickets t
    JOIN technicians tech
        ON t.technician_id = tech.technician_id
    WHERE t.status = 'Resolved'
    GROUP BY tech.technician_name
    ORDER BY resolved_tickets DESC;
    """
)


# ------------------------------------------------------------
# 18. AVERAGE RESOLUTION TIME BY TECHNICIAN
# SQLite calculates time differences in days using julianday().
# Multiply by 24 to convert days into hours.
# ------------------------------------------------------------

avg_resolution_time = run_query(
    "8. AVERAGE RESOLUTION TIME BY TECHNICIAN",
    """
    SELECT
        tech.technician_name,
        ROUND(
            AVG(
                (julianday(t.resolved_date) - julianday(t.created_date)) * 24
            ),
            2
        ) AS avg_resolution_hours
    FROM tickets t
    JOIN technicians tech
        ON t.technician_id = tech.technician_id
    WHERE
        t.status = 'Resolved'
        AND t.resolved_date IS NOT NULL
    GROUP BY tech.technician_name
    ORDER BY avg_resolution_hours;
    """
)


# ------------------------------------------------------------
# 19. AVERAGE RESOLUTION TIME BY PRIORITY
# ------------------------------------------------------------

priority_resolution_time = run_query(
    "9. AVERAGE RESOLUTION TIME BY PRIORITY",
    """
    SELECT
        priority,
        ROUND(
            AVG(
                (julianday(resolved_date) - julianday(created_date)) * 24
            ),
            2
        ) AS avg_resolution_hours
    FROM tickets
    WHERE resolved_date IS NOT NULL
    GROUP BY priority
    ORDER BY avg_resolution_hours;
    """
)


# ------------------------------------------------------------
# 20. CURRENTLY OPEN OR IN-PROGRESS TICKETS
# ------------------------------------------------------------

active_tickets = run_query(
    "10. CURRENT OPEN AND IN-PROGRESS TICKETS",
    """
    SELECT
        t.ticket_id,
        u.first_name || ' ' || u.last_name AS employee,
        d.department_name,
        c.category_name,
        t.issue_description,
        t.priority,
        t.status,
        tech.technician_name,
        t.created_date
    FROM tickets t
    JOIN users u
        ON t.user_id = u.user_id
    JOIN departments d
        ON u.department_id = d.department_id
    JOIN categories c
        ON t.category_id = c.category_id
    LEFT JOIN technicians tech
        ON t.technician_id = tech.technician_id
    WHERE t.status IN ('Open', 'In Progress')
    ORDER BY
        CASE
            WHEN t.priority = 'Critical' THEN 1
            WHEN t.priority = 'High' THEN 2
            WHEN t.priority = 'Medium' THEN 3
            WHEN t.priority = 'Low' THEN 4
        END,
        t.created_date;
    """
)


# ------------------------------------------------------------
# 21. CRITICAL TICKETS
# ------------------------------------------------------------

critical_tickets = run_query(
    "11. CRITICAL-PRIORITY TICKETS",
    """
    SELECT
        ticket_id,
        issue_description,
        status,
        created_date,
        resolved_date
    FROM tickets
    WHERE priority = 'Critical'
    ORDER BY created_date DESC;
    """
)


# ------------------------------------------------------------
# 22. UNRESOLVED CRITICAL TICKETS
# ------------------------------------------------------------

critical_open = run_query(
    "12. UNRESOLVED CRITICAL TICKETS",
    """
    SELECT
        t.ticket_id,
        u.first_name || ' ' || u.last_name AS employee,
        c.category_name,
        t.issue_description,
        tech.technician_name,
        t.status,
        t.created_date
    FROM tickets t
    JOIN users u
        ON t.user_id = u.user_id
    JOIN categories c
        ON t.category_id = c.category_id
    LEFT JOIN technicians tech
        ON t.technician_id = tech.technician_id
    WHERE
        t.priority = 'Critical'
        AND t.status != 'Resolved';
    """
)


# ------------------------------------------------------------
# 23. TICKETS OPEN LONGER THAN 2 DAYS
# We use the project date of 2026-09-22.
# ------------------------------------------------------------

older_tickets = run_query(
    "13. TICKETS OPEN FOR MORE THAN 2 DAYS",
    """
    SELECT
        ticket_id,
        issue_description,
        priority,
        status,
        created_date,
        ROUND(
            julianday('2026-09-22 23:59:59') - julianday(created_date),
            1
        ) AS days_open
    FROM tickets
    WHERE
        status != 'Resolved'
        AND julianday('2026-09-22 23:59:59') - julianday(created_date) > 2
    ORDER BY days_open DESC;
    """
)


# ------------------------------------------------------------
# 24. CATEGORY + PRIORITY BREAKDOWN
# ------------------------------------------------------------

category_priority = run_query(
    "14. CATEGORY AND PRIORITY BREAKDOWN",
    """
    SELECT
        c.category_name,
        t.priority,
        COUNT(*) AS ticket_count
    FROM tickets t
    JOIN categories c
        ON t.category_id = c.category_id
    GROUP BY
        c.category_name,
        t.priority
    ORDER BY
        c.category_name,
        ticket_count DESC;
    """
)


# ------------------------------------------------------------
# 25. FIRST-LEVEL VS ADVANCED TECHNICIAN WORK
# ------------------------------------------------------------

skill_level_summary = run_query(
    "15. TICKETS HANDLED BY TECHNICIAN SKILL LEVEL",
    """
    SELECT
        tech.skill_level,
        COUNT(t.ticket_id) AS total_tickets,
        SUM(
            CASE
                WHEN t.status = 'Resolved' THEN 1
                ELSE 0
            END
        ) AS resolved_tickets
    FROM technicians tech
    LEFT JOIN tickets t
        ON tech.technician_id = t.technician_id
    GROUP BY tech.skill_level
    ORDER BY total_tickets DESC;
    """
)


# ------------------------------------------------------------
# 26. SLA ANALYSIS
#
# Example SLA:
# Critical = 2 hours
# High = 4 hours
# Medium = 8 hours
# Low = 24 hours
# ------------------------------------------------------------

sla_analysis = run_query(
    "16. SLA PERFORMANCE ANALYSIS",
    """
    WITH resolution_data AS (
        SELECT
            ticket_id,
            priority,
            ROUND(
                (julianday(resolved_date) - julianday(created_date)) * 24,
                2
            ) AS resolution_hours
        FROM tickets
        WHERE resolved_date IS NOT NULL
    )

    SELECT
        ticket_id,
        priority,
        resolution_hours,

        CASE
            WHEN priority = 'Critical' AND resolution_hours <= 2
                THEN 'Met SLA'
            WHEN priority = 'High' AND resolution_hours <= 4
                THEN 'Met SLA'
            WHEN priority = 'Medium' AND resolution_hours <= 8
                THEN 'Met SLA'
            WHEN priority = 'Low' AND resolution_hours <= 24
                THEN 'Met SLA'
            ELSE 'Missed SLA'
        END AS sla_status

    FROM resolution_data
    ORDER BY ticket_id;
    """
)


# ------------------------------------------------------------
# 27. SLA SUMMARY
# ------------------------------------------------------------

sla_summary = run_query(
    "17. SLA SUMMARY",
    """
    WITH resolution_data AS (
        SELECT
            ticket_id,
            priority,
            (julianday(resolved_date) - julianday(created_date)) * 24
                AS resolution_hours
        FROM tickets
        WHERE resolved_date IS NOT NULL
    ),

    sla_results AS (
        SELECT
            ticket_id,
            priority,
            resolution_hours,

            CASE
                WHEN priority = 'Critical' AND resolution_hours <= 2
                    THEN 'Met SLA'
                WHEN priority = 'High' AND resolution_hours <= 4
                    THEN 'Met SLA'
                WHEN priority = 'Medium' AND resolution_hours <= 8
                    THEN 'Met SLA'
                WHEN priority = 'Low' AND resolution_hours <= 24
                    THEN 'Met SLA'
                ELSE 'Missed SLA'
            END AS sla_status

        FROM resolution_data
    )

    SELECT
        sla_status,
        COUNT(*) AS ticket_count
    FROM sla_results
    GROUP BY sla_status;
    """
)


# ------------------------------------------------------------
# 28. WINDOW FUNCTION:
# RANK TECHNICIANS BY NUMBER OF RESOLVED TICKETS
# ------------------------------------------------------------

technician_rank = run_query(
    "18. TECHNICIAN PERFORMANCE RANKING",
    """
    WITH technician_stats AS (
        SELECT
            tech.technician_name,
            COUNT(t.ticket_id) AS resolved_tickets
        FROM technicians tech
        LEFT JOIN tickets t
            ON tech.technician_id = t.technician_id
            AND t.status = 'Resolved'
        GROUP BY tech.technician_name
    )

    SELECT
        technician_name,
        resolved_tickets,
        RANK() OVER (
            ORDER BY resolved_tickets DESC
        ) AS performance_rank
    FROM technician_stats;
    """
)


# ------------------------------------------------------------
# 29. DAILY TICKET VOLUME
# ------------------------------------------------------------

daily_volume = run_query(
    "19. DAILY TICKET VOLUME",
    """
    SELECT
        DATE(created_date) AS ticket_date,
        COUNT(*) AS tickets_created
    FROM tickets
    GROUP BY DATE(created_date)
    ORDER BY ticket_date;
    """
)


# ------------------------------------------------------------
# 30. CREATE VISUALIZATIONS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CREATING CHARTS")
print("=" * 70)


# Chart 1: Tickets by Status
plt.figure(figsize=(8, 5))
plt.bar(
    status_summary["status"],
    status_summary["ticket_count"]
)
plt.title("Help Desk Tickets by Status")
plt.xlabel("Status")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=20)
plt.show()


# Chart 2: Tickets by Priority
plt.figure(figsize=(8, 5))
plt.bar(
    priority_summary["priority"],
    priority_summary["ticket_count"]
)
plt.title("Tickets by Priority")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.show()


# Chart 3: Most Common Issue Categories
plt.figure(figsize=(10, 6))
plt.barh(
    category_summary["category_name"],
    category_summary["ticket_count"]
)
plt.title("Most Common IT Support Issues")
plt.xlabel("Number of Tickets")
plt.ylabel("Issue Category")
plt.gca().invert_yaxis()
plt.show()


# Chart 4: Technician Workload
plt.figure(figsize=(10, 6))
plt.bar(
    technician_workload["technician_name"],
    technician_workload["assigned_tickets"]
)
plt.title("Tickets Assigned by Technician")
plt.xlabel("Technician")
plt.ylabel("Assigned Tickets")
plt.xticks(rotation=30)
plt.show()


# Chart 5: Tickets by Department
plt.figure(figsize=(10, 6))
plt.bar(
    department_summary["department_name"],
    department_summary["ticket_count"]
)
plt.title("Support Tickets by Department")
plt.xlabel("Department")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=30)
plt.show()


# Chart 6: Daily Ticket Volume
plt.figure(figsize=(12, 5))
plt.plot(
    daily_volume["ticket_date"],
    daily_volume["tickets_created"],
    marker="o"
)
plt.title("Daily Help Desk Ticket Volume")
plt.xlabel("Date")
plt.ylabel("Tickets Created")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------------------
# 31. PROJECT KPI SUMMARY
# ------------------------------------------------------------

kpi_query = """
SELECT
    COUNT(*) AS total_tickets,

    SUM(
        CASE
            WHEN status = 'Resolved'
            THEN 1
            ELSE 0
        END
    ) AS resolved_tickets,

    SUM(
        CASE
            WHEN status != 'Resolved'
            THEN 1
            ELSE 0
        END
    ) AS unresolved_tickets,

    SUM(
        CASE
            WHEN priority = 'Critical'
            THEN 1
            ELSE 0
        END
    ) AS critical_tickets,

    ROUND(
        AVG(
            CASE
                WHEN resolved_date IS NOT NULL
                THEN
                    (julianday(resolved_date) - julianday(created_date)) * 24
            END
        ),
        2
    ) AS avg_resolution_hours

FROM tickets;
"""

kpis = pd.read_sql_query(kpi_query, conn)

print("\n" + "=" * 70)
print("HELP DESK KPI SUMMARY")
print("=" * 70)

display(kpis)


# ------------------------------------------------------------
# 32. EXPORT ANALYSIS RESULTS TO CSV FILES
# ------------------------------------------------------------

status_summary.to_csv("ticket_status_summary.csv", index=False)
priority_summary.to_csv("ticket_priority_summary.csv", index=False)
category_summary.to_csv("issue_category_summary.csv", index=False)
department_summary.to_csv("department_ticket_summary.csv", index=False)
technician_workload.to_csv("technician_workload.csv", index=False)
avg_resolution_time.to_csv("technician_resolution_times.csv", index=False)
sla_analysis.to_csv("sla_analysis.csv", index=False)
daily_volume.to_csv("daily_ticket_volume.csv", index=False)

print("\nCSV analysis files exported successfully.")


# ------------------------------------------------------------
# 33. DISPLAY DATABASE TABLE NAMES
# ------------------------------------------------------------

tables = pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
""", conn)

print("\n" + "=" * 70)
print("DATABASE TABLES")
print("=" * 70)

display(tables)


# ------------------------------------------------------------
# 34. FINAL PROJECT MESSAGE
# ------------------------------------------------------------

print("""
============================================================
PROJECT COMPLETE
============================================================

You created a Help Desk Ticket System using:

- SQL
- SQLite
- Google Colab
- Python
- Pandas
- Matplotlib

SQL skills demonstrated:

- SELECT
- WHERE
- JOIN
- GROUP BY
- ORDER BY
- COUNT
- AVG
- CASE
- CTEs
- Date calculations
- Window functions
- RANK()
- SLA analysis

Portfolio Project:
Help Desk Ticket System & IT Support Analytics

Next step:
Save this Google Colab notebook and upload it to GitHub.
============================================================
""")
```
